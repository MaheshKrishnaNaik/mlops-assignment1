# 01 - Exploratory Data Analysis
Heart Disease UCI (Cleveland). Run `python -m src.data` first.

```python
import sys
sys.path.insert(0, '.')
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from src import config
from src.data import load_clean
df = load_clean()
print(df.shape)
df.head()
```

```python
df.describe().T
```

## Class balance

```python
sns.countplot(x=config.TARGET, data=df)
plt.title('Class balance')
print(df[config.TARGET].value_counts(normalize=True))
```

## Numeric distributions

```python
df[config.NUMERIC].hist(bins=20, figsize=(12, 7))
plt.show()
```

## Correlation heatmap

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.show()
```

## Categorical features versus target

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ['cp', 'thal', 'ca']):
    sns.countplot(x=col, hue=config.TARGET, data=df, ax=ax)
plt.show()
```
